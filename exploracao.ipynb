```python
import os
import sys

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

from pyspark.sql import SparkSession

spark = (SparkSession.builder
    .appName("EngordAI-MultiTable")
    .master("local[*]")
    .config("spark.driver.memory", "4g")
    .config("spark.driver.host", "127.0.0.1")
    .config("spark.driver.bindAddress", "127.0.0.1")
    .config("spark.python.worker.reuse", "false")
    .getOrCreate())

# Importa o módulo
from gerador_cdb import generate_cdb_tables, validate_relationships

tipo_if, instrumento, operacao = generate_cdb_tables(
    spark,
    n_instrumentos=1000,
    n_operacoes=50000,
    seed=42,
)

tipo_if.show()
instrumento.show(5)
operacao.show(5)

# Valida
validate_relationships(tipo_if, instrumento, operacao)

tipo_if.toPandas().to_csv("data_csv/tipo_if.csv", index=False)
instrumento.toPandas().to_csv("data_csv/instrumento.csv", index=False)
operacao.toPandas().to_csv("data_csv/operacao.csv", index=False)
```
